# Building a Todo Task Manager Agent

This tutorial shows how to build a practical agent that manages tasks using the Todoist API.

## What This Agent Does:
- Lists your projects and tasks
- Completes tasks
- Updates task due dates
- Deletes tasks
- Provides intelligent task management based on natural language commands

## Steps:
1. Set up Todoist API credentials
2. Define Todo actions (direct API wrappers)
3. Create the TodoAgent
4. Add examples to guide the agent
5. Test with real tasks

**Note:** This is a practical implementation - no environment wrapper or reward system needed!

## Setup: Install Dependencies and Configure API

````python
# Install required packages if needed
# !pip install requests python-dotenv agentlite

import os
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# You need to set your Todoist API token
# Get it from: https://todoist.com/app/settings/integrations/developer
# Add to your .env file: TODO_KEY=your_token_here
TODO_API_KEY = os.getenv("TODO_KEY")

if not TODO_API_KEY:
    print("⚠️  Warning: TODO_KEY not found in environment variables")
    print("Please add your Todoist API token to .env file")
else:
    print("✅ Todoist API key loaded successfully")
````

## Create the Todo Manager Agent

Import the agent class from the module instead of defining inline!

````python
from agentlite.llm.agent_llms import get_llm_backend
from agentlite.llm.LLMConfig import LLMConfig
from todo_actions.todo_agent import TodoManagerAgent

print("✅ TodoManagerAgent class imported from todo_actions.todo_agent module")
````

## Initialize the Agent with LLM

````python
# Configure LLM (adjust based on your setup)
llm_config_dict = {
    "llm_name": os.getenv("LLM", "gpt-3.5-turbo"),
    "temperature": 0.0
}

# If using OpenRouter
if os.getenv("OPENROUTER_API_KEY"):
    llm_config_dict["provider"] = "openrouter"
    llm_config_dict["api_key"] = os.getenv("OPENROUTER_API_KEY")
    llm_config_dict["base_url"] = os.getenv("OPENROUTER_API_BASE")

llm_config = LLMConfig(llm_config_dict)
llm = get_llm_backend(llm_config)

# Create the agent
todo_agent = TodoManagerAgent(llm=llm)

print("✅ Agent initialized and ready to use!")
````

## Test the Agent

Now let's test with real commands!

````python
from agentlite.commons import TaskPackage

# Test 1: List all projects
print("=" * 60)
print("TEST 1: List all my projects")
print("=" * 60)

task1 = "Show me all my projects"
task_pack1 = TaskPackage(instruction=task1)
response1 = todo_agent(task_pack1)
print(f"\n📋 Response: {response1}\n")
````

````python
# Test 2: Get tasks in a specific project
print("=" * 60)
print("TEST 2: Get tasks in Inbox project")
print("=" * 60)

task2 = "What tasks do I have in my Inbox?"
task_pack2 = TaskPackage(instruction=task2)
response2 = todo_agent(task_pack2)
print(f"\n📝 Response: {response2}\n")
````

````python
# Test 3: Update a task's due date
print("=" * 60)
print("TEST 3: Update task due date")
print("=" * 60)

# Note: You'll need to replace with an actual task ID from your Todoist
task3 = "Change the due date of task ID 123456789 to tomorrow"
task_pack3 = TaskPackage(instruction=task3)
response3 = todo_agent(task_pack3)
print(f"\n📅 Response: {response3}\n")
````

## Modular Structure vs Benchmark Code

This tutorial uses a **modular structure** similar to benchmarks, but without unnecessary complexity:

### 📁 File Structure (Like Benchmarks):
```
tutorials/
  ├── todo_task_manager.ipynb     # This notebook
  └── todo_actions/                # Action module (like benchmark/tool-operation)
      ├── __init__.py              # Package exports
      ├── todo_actions.py          # Action implementations
      └── todo_agent.py            # Agent definition
```

### ❌ What We DON'T Have (Benchmark-only features):
- **No `todo_env.py`**: No `TodoEnv` environment class
- **No `step()` Function**: Actions execute directly via API
- **No Reward System**: No scoring or ground truth comparison
- **No Subgoal Tracking**: No need to measure partial completion
- **No `action_path` tracking**: Simpler execution model

### ✅ What We DO Have (Practical + Modular):
- **Modular Structure**: Organized like benchmarks for maintainability
- **Direct API Calls**: Actions call Todoist API directly
- **Reusable Components**: Import actions into any notebook
- **Clean Separation**: Actions, agent, and usage are separated
- **Easy to Maintain**: Much less code complexity than benchmark

### When to Use Each Pattern:

**Use Benchmark Pattern (with Environment + Rewards) when:**
- Comparing different agent architectures
- Measuring agent performance quantitatively
- Research and development
- Need reproducible evaluation metrics

**Use Practical Pattern (modular, no environment) when:**
- Building production applications
- User-facing tools
- Real-world task automation
- Simplicity and maintainability matter
- Want to reuse actions across projects

This tutorial demonstrates the **practical modular pattern** - organized structure without unnecessary complexity! 🎯
